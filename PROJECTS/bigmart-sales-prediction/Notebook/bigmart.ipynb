### Building Model

### Base Line

```python
import numpy as np  
from sklearn.dummy import DummyRegressor
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import (
    r2_score, mean_squared_error, mean_absolute_error,
)


# 1. Select features and target


features = [
    'Item_Weight', 'Item_Visibility', 'Item_MRP',
    'Item_Fat_Content', 'Item_Type', 'Outlet_Size',
    'Outlet_Location_Type', 'Outlet_Type'
]

X = df[features]
y = df_clean['Item_Outlet_Sales_Logged']

# 2. Split AFTER selecting the features
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=0
)

# 3. Define preprocessing
numeric_features = [
    'Item_Weight', 'Item_Visibility', 'Item_MRP'
]

categorical_features = [
    'Item_Fat_Content', 'Item_Type', 'Outlet_Size',
    'Outlet_Location_Type', 'Outlet_Type'
]

numeric_transformer = SimpleImputer(strategy='median')

categorical_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore'))
])

preprocess = ColumnTransformer([
    ('num', numeric_transformer, numeric_features),
    ('cat', categorical_transformer, categorical_features)
])

```

```python

dummy = DummyRegressor(strategy='mean')
dummy.fit(y_train.to_frame(), y_train)

dummy=DummyRegressor(strategy='mean')
dummy.fit(X_train, y_train)
dummy_pred = dummy.predict(X_test)

print("Dummy Model Score (R^2): {:.3f}".format(dummy.score(X_test, y_test)))

```

```python
from sklearn.linear_model import LinearRegression

model = Pipeline([
    ('processing', preprocess),
    ('model', LinearRegression())
])

model.fit(X_train,y_train)
cross=cross_val_score(model, X_train,y_train, cv=5, scoring='r2')

#On logged scaled data
print("Best R2 {:.3f}".format(cross.mean()))
print("R2 train {:.3f}".format(model.score(X_train,y_train)))
print("R2 test {:.3f}".format(model.score(X_test,y_test)))
```

```python
y_pred=model.predict(X_test)

y_pred_original=np.expm1(y_pred)
y_test_original=np.expm1(y_test)

print("R2 (Original Scale): {:3f}".format(r2_score(y_test_original, y_pred_original)))
print("Mean Squared Error (Original Scale): {:.3f}".format(mean_squared_error(y_test_original, y_pred_original)))
print("Root Mean Squared Error (Original Scale): {:.3f}".format(np.sqrt(mean_squared_error(y_test_original, y_pred_original))))
print("Mean Absolute Error (Original Scale): {:.3f}".format(mean_absolute_error(y_test_original, y_pred_original)))

```

```python
#Actual plot vs predicted plot and residual
plt.figure(figsize=(10, 6))
sns.scatterplot(x=y_test_original, y=y_pred_original, alpha=0.5)

lower = min(y_test_original.min(), y_pred_original.min())
upper = max(y_test_original.max(), y_pred_original.max())
plt.plot([lower, upper], [lower, upper], "r--", label="Perfect prediction")

plt.xlabel("Actual Sales")
plt.ylabel("Predicted Sales")
plt.title("Actual vs Predicted Sales")
plt.tight_layout()
plt.show()
```

### Calculating comparable test metrics

```python
import numpy as np
import pandas as pd

from sklearn.metrics import (
    r2_score,
    mean_absolute_error,
    mean_squared_error
)
best_models={
    'Decision Tree':best_estimator_deci,
    'Random Forest':best_estimator_rand,
    'Gradient Boosting':best_est_Gra

}

results = []

y_true = np.expm1(np.asarray(y_test))

for name, model in best_models.items():
    predictions = model.predict(X_test)


    predictions = np.expm1(np.asarray(predictions))
    

    mae = mean_absolute_error(y_true, predictions)
    mse = mean_squared_error(y_true, predictions)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_true, predictions)

    results.append({
        "Model": name,
        "R2": r2,
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse
    })

results_df = pd.DataFrame(results)
results_df = results_df.sort_values(
    by="RMSE",
    ascending=True
).reset_index(drop=True)

print(results_df.round(3))
```
