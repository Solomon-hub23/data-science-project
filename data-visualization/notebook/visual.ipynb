### Our primary focus is on `cnt` (the number of total rentals per day) and we would visualize these business questions:

- What does rental demand look like?
- Does temperature influence demand?
- Do working days matter?
- Does weather impact rentals?


```python
plt.figure(figsize=(10,5))
plt.plot(df['cnt'])

plt.grid(True)
plt.title("Total Rental Over time")
plt.xlabel("Number of days")
plt.ylabel("Total Rental")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df['temp'], df['cnt'], alpha=0.6)
plt.title("Temperature Vs Total Rentals", fontsize=16)
plt.xlabel("Temperature", fontsize=13)
plt.ylabel("Total Rentals", fontsize=13)

plt.xticks(fontsize=11)
plt.yticks(fontsize=11)
plt.grid(True)
plt.show()
```

```python
working_day=df[df['workingday']==1]
non_working_day=df[df["workingday"]==0]

plt.figure(figsize=(9,6))
plt.subplot(2,1,1)
plt.plot(working_day['cnt'])
plt.title("Rental during working days over time")
plt.xlabel("Days")
plt.ylabel("total Rentals(working days)")
plt.grid(True)

plt.subplot(2,1,2)
plt.plot(non_working_day['cnt'])

plt.title("Rental during non working day over time")
plt.xlabel("Days")
plt.ylabel("total Rentals(Non working days)")
plt.tight_layout()
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(9,6))


plt.subplot(2,1,1)
plt.hist(working_day['cnt'], bins=30, edgecolor="black")
plt.title("Distribution of Rental for working days")
plt.xlabel("Working Day")

plt.subplot(2,1,2)

plt.hist(non_working_day['cnt'], bins=30, edgecolor="black")
plt.title("Distribution Rental for non-working days")
plt.xlabel("Non-Working Day")

plt.tight_layout()
plt.show()
```

### Now let us consider looking out for any outliers and the spread of total rental `cnt`

```python

plt.figure(figsize=(8, 3))

plt.boxplot(df["cnt"],
            vert=False)

plt.title("Boxplot of Daily Bike Rentals", fontsize=14)
plt.xlabel("Total Daily Rentals", fontsize=12)

plt.yticks([])  
plt.grid(axis="x", linestyle="--", alpha=0.6)

plt.show()
```

### Interpreting the Boxplot
The median daily rental count is located roughly around the mid-4,000 range, indicating that a typical day sees approximately 4,500 rentals.

The interquartile range (the width of the box) appears to span roughly from around 3,000 to 6,000 rentals. This means that 50% of all observed days fall within this range, confirming moderate variability in daily demand.

The whiskers extend to very low and very high values, indicating that rental counts can vary substantially across days. However, there are no isolated outlier markers shown in this plot, suggesting that extreme values are not separated from the overall distribution.

Overall, the boxplot confirms that daily demand is centered around a stable median but exhibits meaningful dispersion. This variability implies that operational decisions — such as staffing levels and bike allocation — should remain flexible to accommodate fluctuations in demand.

```python

plt.figure(figsize=(8, 3))

plt.boxplot(df['temp'], vert=False)
plt.xlabel("Temperature")
plt.yticks([])

plt.grid(axis="x", linestyle="--", alpha=0.6)
plt.show()
```

The temperature is centered around 0.50 with IQR of ~0.34 to ~0.66 normalized(-11 degree C to 23 degree c), meaning that 50% of observations fall within this moderate range.

The distribution is slightly left-skewed with whiskers extending from ~0.06 to ~0.86, indicating high variability across the year but no significant outliers.
