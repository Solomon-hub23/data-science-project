### DATA CLEANING 

```python
Airbnb_data_clean=Airbnb_data.copy()
```

```python
Airbnb_data_clean.columns=(
    Airbnb_data_clean.columns.str.strip().str.lower().str.replace(" ","_")
)
Airbnb_data_clean.columns
```

```python
Airbnb_data_clean.isnull().sum()
```

```python
Airbnb_data_clean.columns
```

```python
Airbnb_data_clean['instant_bookable']=Airbnb_data_clean['instant_bookable'].astype(bool)
Airbnb_data_clean['last_review']=pd.to_datetime(Airbnb_data_clean['last_review'])

#Checking to see whether the following attributes have off data values
cols=['calculated_host_listings_count','review_rate_number','minimum_nights', 'construction_year','availability_365','number_of_reviews']
for col in cols:
    print(f"\n======{col}=======")
    print(Airbnb_data_clean[col].unique())
```

```python
for col in cols:
    Airbnb_data_clean[col]=Airbnb_data_clean[col].astype("Int64")
```

```python

cols=['price','service_fee']
for col in cols:
    Airbnb_data_clean[col]=Airbnb_data_clean[col].replace(r"[$]"," ", regex=True).replace(r"[\,]","", regex=True)
    Airbnb_data_clean[col]=Airbnb_data_clean[col].astype("Int64")
```

```python

missing_values=Airbnb_data_clean.isnull().mean()*100
print(missing_values[missing_values>0].sort_values(ascending=False))
```

```python
Airbnb_data_clean=Airbnb_data_clean.drop(['license'], axis=1)
```
