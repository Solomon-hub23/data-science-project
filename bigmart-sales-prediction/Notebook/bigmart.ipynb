### Improved Features

```python

from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import (
    r2_score, mean_squared_error, mean_absolute_error
)
import numpy as np

# 1. Select features and target
features = [
    'Item_Weight', 'Item_Visibility', 'Item_MRP',
    'Item_Fat_Content', 'Item_Type', 'Outlet_Size',
    'Outlet_Location_Type', 'Outlet_Type'
]

X = df[features]
y = df['Item_Outlet_Sales']

# 2. Split AFTER selecting the features
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=0
)

# 3. Define preprocessing
numeric_features = [
    'Item_Weight', 'Item_Visibility', 'Item_MRP'
]

categorical_features = [
    'Item_Fat_Content', 'Item_Type', 'Outlet_Size',
    'Outlet_Location_Type', 'Outlet_Type'
]

numeric_transformer = SimpleImputer(strategy='median')

categorical_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore'))
])

preprocess = ColumnTransformer([
    ('num', numeric_transformer, numeric_features),
    ('cat', categorical_transformer, categorical_features)
])

# 4. Build the complete model
model = Pipeline([
    ('processing', preprocess),
    ('model', DecisionTreeRegressor(
        max_depth=10,
        min_samples_split=50,
        min_samples_leaf=50,
        random_state=0
    ))
])

# 5. Cross-validation on the training set
cross = cross_val_score(
    model, X_train, y_train, cv=cv, scoring='r2'
)

print(
    "Cross-validation R²: {:.3f} ({:.3f})".format(
        cross.mean(), cross.std()
    )
)

# 6. Fit on the full training set
model.fit(X_train, y_train)

# 7. Predict on the test set
y_pred = model.predict(X_test)

# 8. Convert back to original sales scale IF y is log-transformed
y_test_original = np.expm1(y_test)
y_pred_original = np.expm1(y_pred)

# 9. Evaluate on the original sales scale
print("Test on original scale R²: {:.3f}".format(
    r2_score(y_test_original, y_pred_original)
))
print("Test MSE: {:.3f}".format(
    mean_squared_error(y_test_original, y_pred_original)
))
print("Test RMSE: {:.3f}".format(
    np.sqrt(mean_squared_error(y_test_original, y_pred_original))
))
print("Test MAE: {:.3f}".format(
    mean_absolute_error(y_test_original, y_pred_original)
))

y_pred = model.predict(X_test)
print("R2 on log-scaled:{:.3f}".format(r2_score(y_test, y_pred)))
```
